# Lists and Dictionaries with For Loops, List Comprehension, and Dictionary Comprehension

```python
# Looping through lists
numbers = [10, 20, 30, 40, 50]
total = 0

for num in numbers:
    total += num

print("Total sum:", total)
```

```python
# Doubling each number in a list
numbers = [1, 2, 3, 4, 5]
doubled = []

for num in numbers:
    doubled.append(num * 2)

print("Doubled List:", doubled)
```

```python
# Example
foods = ["Dosa", "Idli", "Vada", "Bisibelebath"]

for food in foods:
    print(f"I like {food}")
```

```python
# Looping through dictionaries
#Example: Iterating over dictionary keys
student_marks = {"Anand": 85, "Geetha": 90, "Kumar": 78}

for student in student_marks:
    print(student)
```

```python
#Example: Iterating over dictionary values
student_marks = {"Anand": 85, "Geetha": 90, "Kumar": 78}

for marks in student_marks.values():
    print(marks)
```

```python
# Both keys and values
student_marks = {"Anand": 85, "Geetha": 90, "Kumar": 78}

for student, marks in student_marks.items():
    print(f"{student} scored {marks} marks")
```

```python
# for loop with range()
#Example: Adding marks to students using index values
students = ["Anand", "Geetha", "Kumar"]
marks = [85, 90, 78]

student_marks = {}

for i in range(len(students)):
    student_marks[students[i]] = marks[i]

print(student_marks)
```

```python
#List comprehension
# Syntax : new_list = [expression for item in iterable if condition]
```

```python
# Ex : squaring numbers in a list
numbers = [1, 2, 3, 4, 5]
squares = [num ** 2 for num in numbers]
print(squares)
```

```python
# Ex: Filtering even numbers
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = [num for num in numbers if num % 2 == 0]
print(even_numbers)
```

```python
# Ex : Uppercasing city names
cities = ["Bengaluru", "Mysuru", "Hubballi", "Mangaluru"]
uppercased_cities = [city.upper() for city in cities]
print(uppercased_cities)
```

```python
# Dictionary Comprehension
# Syntax : new_dict = {key_expression: value_expression for item in iterable if condition}
```

```python
# Ex : Creating dictionary of squares
numbers = [1, 2, 3, 4, 5]
squares_dict = {num: num ** 2 for num in numbers}
print(squares_dict)
```

```python
# Ex ; Converting a list of names to a dictionary of name lengths
names = ["Anand", "Geetha", "Kumar"]
name_lengths = {name: len(name) for name in names}
print(name_lengths)
```

```python
# Ex : Filtering cities with population above 10 lakhs
city_population = {
    "Bengaluru": 84,
    "Mysuru": 11,
    "Hubballi": 9,
    "Mangaluru": 5
}
large_cities = {city: population for city, population in city_population.items() if population > 10}
print(large_cities)
```

```python
# Splitting strings to create lists
# syntax : string.split(separator, maxsplit)
```

```python
# Ex : Splitting sentence into words
sentence = "I love coding in Python"
words = sentence.split()
print(words)
```

```python
# Ex : Splitting with comas 
data = "apple,banana,mango"
fruits = data.split(",")
print(fruits)
```

```python
# Ex: LImiting the number of splits
sentence = "Python is fun to learn"
words = sentence.split(" ", 2)
print(words)
```
