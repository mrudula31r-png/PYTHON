# Lists in python

```python
# What is list ?       A list is a collection of items that are ordered, mutable (changeable), 
#          and allow duplicate elements. Lists can hold items of different data types, 
#          such as integers, strings, or even other lists.
# Syntax : my_list = [element1, element2, element3, ...]
```

```python
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = ["apple", 3, True]
print(fruits)
print(numbers)
print(mixed)
```

```python
# Accessing List - Syntax : list_name[index]
fruits = ["apple", "banana", "cherry"]
print(fruits[0])  
print(fruits[2])  
```

```python
# Negative indexing
print(fruits[-1])  
print(fruits[-2])  
```

```python
# Modifying Lists
fruits[1] = "orange"
print(fruits)  
```

```python
# Append means adds elements at the end
fruits.append("grape")
print(fruits)  
```

```python
 # Insert means an element is inserted at a specific index
fruits.insert(1, "kiwi")
print(fruits)  
```

```python
# Remove : removes element
fruits.remove("kiwi")
print(fruits)  # Output: ['apple', 'kiwi', 'cherry']
```

```python
# Pop: Removes the element at a specific index (or the last item if no index is provided).
fruits.pop()  
print(fruits)  

fruits.pop(0)  
print(fruits)  
```

```python
# Clear : Removes all elements from the list.
fruits.clear()
print(fruits)  
```

```python
# Slicing Lists :You can extract a portion of a list using slicing.
# Syntax : list_name[start:stop:step] 
numbers = [0, 1, 2, 3, 4, 5, 6]
print(numbers[1:4])  
print(numbers[:4])  
print(numbers[2:])  
print(numbers[::2])  

```

```python
# List Functions and Methods
# len(list): Returns the number of elements in the list.
fruits = ["apple", "banana", "cherry"]
print(len(fruits)) 
```

```python
# sorted(list): Returns a new sorted list without changing the original list.
numbers = [5, 2, 9, 1]
print(sorted(numbers))  
print(numbers)
```

```python
#sum(list): Returns the sum of elements in a list (for numerical lists).
numbers = [1, 2, 3, 4]
print(sum(numbers))  
```

```python
# index(element): Returns the index of the first occurrence of the specified element.
print(fruits.index("apple"))  
```

```python
# count(element): Returns the number of occurrences of an element in the list.
numbers = [1, 2, 3, 1, 1]
print(numbers.count(1))  
```

```python
# reverse(): Reverses the elements of the list in place.
fruits.reverse()
print(fruits)  
```

```python
# sort(): Sorts the list in place (ascending by default).
numbers = [5, 2, 9, 1]
numbers.sort()
print(numbers)  
```

```python
#      Nested Lists : Lists can contain other lists, allowing you to create nested lists.
#     This can be useful for storing matrix-like data structures.
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Accessing elements in a nested list
print(matrix[0])  
print(matrix[1][1])  

```
