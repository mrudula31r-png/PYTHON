```python
# Strings Methods
message = "  Hello World!  "
print(message.strip())  
print(message.upper())  
print(message.lower())
print(message.replace("World", "Python"))  
print(message.replace(" ", "_"))
```

```python
# Acessing string character - Indexing
text = "Python"
print(text[0])  
print(text[2]) 
# Negative indexing
print(text[-1])  
print(text[-3])  
```
